                    **Password Strength Estimation**
A Password Strength Estimation system evaluates how secure a password is and classifies it as Weak, Medium, or Strong.

A strong password usually has:

✔ At least 8 characters

✔ Uppercase letters (A–Z)

✔ Lowercase letters (a–z)

✔ Numbers (0–9)

✔ Special symbols (! @ # $ % & *)

Suppose a password contains only lowercase letters and the password length is 4. There are 26 possible characters, and with a length of 4, total possible passwords are  26⁴ = 456,976. An attacker would have to try up to 456,976 combinations.

Now suppose the password contains lowercase letters, uppercase letters, digits, and special symbols. Total characters available would be 26 + 26 + 10 + 20 = 82 characters. For an 8-character password, possible passwords are 82⁸ ≈ 2.04 × 10¹⁵. This is more than 2 quadrillion possible passwords.

```python
password = input("Enter your password: ")
```

```python
score = 0

# Check length
if len(password) >= 8:
    score += 1
```

```python
# Check uppercase
if any(ch.isupper() for ch in password):
    score += 1
```

```python
# Check lowercase
if any(ch.islower() for ch in password):
    score += 1
```

```python
# Check digits
if any(ch.isdigit() for ch in password):
    score += 1
```

```python
# Check special characters
if any(not ch.isalnum() for ch in password):
    score += 1
```

```python
print("\nPassword Score:", score)

if score <= 2:
    print("Password Strength: Weak")
elif score <= 4:
    print("Password Strength: Medium")
else:
    print("Password Strength: Strong")
```

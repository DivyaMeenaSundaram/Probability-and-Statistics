**Disease Risk Estimation or Disease Prediction Based on Symptoms**

Computers can also assist by estimating the likelihood of a disease based on previously observed symptom patterns.Here is a simple Python program that will estimate the probability of a disease using symptom frequencies. **It does not make a medical diagnosis.**

| Average Probability | Risk Level    |
| ------------------: | ------------- |
|               0–39% | Low Risk      |
|              40–69% | Moderate Risk |
|             70–100% | High Risk     |


```python
symptom_probability = {
    "fever": 95,
    "headache": 80,
    "joint pain": 85,
    "skin rash": 60,
    "fatigue": 70
}
```

```python
print("Available Symptoms")
for symptom in symptom_probability:
    print("-", symptom)
```

```python
print("\nEnter symptoms one by one.")
print("Type 'done' when finished.\n")

selected = []

while True:
    symptom = input("Enter symptom: ").lower()

    if symptom == "done":
        break

    selected.append(symptom)
```

```python
total = 0
count = 0

for symptom in selected:

    if symptom in symptom_probability:
        total += symptom_probability[symptom]
        count += 1

if count == 0:
    print("\nNo matching symptoms found.")
else:

    risk = total / count

    print("\nEstimated Disease Risk = {:.2f}%".format(risk))

    if risk < 40:
        print("Risk Level : LOW")
    elif risk < 70:
        print("Risk Level : MODERATE")
    else:
        print("Risk Level : HIGH")
```

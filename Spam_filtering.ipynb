                    **Application 1: Spam Filtering**

**Spam** refers to unwanted or unsolicited messages that are sent to many people. These messages are usually advertisements, scams, or attempts to trick users into clicking malicious links or sharing personal information.

Eg: Congratulations! You have won ₹10,00,000.
Click here to claim your prize.

**Ham** simply means a legitimate (normal) message. It is not an abbreviation—it's just a term adopted as the opposite of "Spam." These are genuine emails sent by friends, teachers, colleagues, or organizations.

Eg: Meeting starts at 10 AM tomorrow.

```python
emails = [
    ("Win money now", "Spam"),
    ("Claim your free prize", "Spam"),
    ("Congratulations you won", "Spam"),
    ("Free lottery ticket", "Spam"),
    ("Exclusive offer just for you", "Spam"),

    ("Meeting at 10 AM", "Ham"),
    ("Project report submitted", "Ham"),
    ("Lunch with friends", "Ham"),
    ("Assignment due tomorrow", "Ham"),
    ("Happy birthday wishes", "Ham")
]
```

```python
spam_words = {}
ham_words = {}
```

```python
# Count word frequencies
for email, label in emails:

    words = email.lower().split()

    for word in words:

        if label == "Spam":
            spam_words[word] = spam_words.get(word, 0) + 1
        else:
            ham_words[word] = ham_words.get(word, 0) + 1
```

```python
print("Spam Word Frequency")
print(spam_words)

print("\nHam Word Frequency")
print(ham_words)
```

```python
# Test a new email
test_email = input("\nEnter a new email: ").lower().split()
```

```python
spam_score = 0
ham_score = 0

for word in test_email:

    spam_score += spam_words.get(word, 0)
    ham_score += ham_words.get(word, 0)

print("\nSpam Score :", spam_score)
print("Ham Score  :", ham_score)

if spam_score > ham_score:
    print("\nPrediction : SPAM")
else:
    print("\nPrediction : HAM")
```

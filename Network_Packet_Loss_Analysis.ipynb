**Network Packet Loss Analysis**

When data is transmitted over a computer network (such as the Internet), it is divided into small units called packets. Each packet travels independently from the sender to the receiver. Normally, all packets should reach their destination. However, due to network congestion, hardware failures, or weak wireless signals, some packets may be lost during transmission. This is called Packet Loss.

A packet is a small block of data sent across a network. For example, if you send a photo through WhatsApp, the photo is divided into many small packets. The receiver combines all the packets to reconstruct the original file.

Some common reasons for packet loss are:
Network congestion, Weak Wi-Fi signal, Damaged network cables, Router failure,
Signal interference

High packet loss can cause:
Video buffering, Poor video call quality, Audio interruptions, Online gaming lag

Suppose

Packets Sent = 100
Packets Received = 95

Then, lost Packets: 100 − 95 = 5

1.png

This means there is a 5% probability that a packet was lost during transmission.

Decision Rule
| Packet Loss | Network Quality |
| ----------- | --------------- |
| 0–2%        | Excellent       |
| 2–5%        | Good            |
| Above 5%    | Poor            |


```python
sent = int(input("Enter packets sent: "))
received = int(input("Enter packets received: "))
```

```python
if received > sent:
    print("Invalid input! Received packets cannot exceed sent packets.")
else:
    lost = sent - received
    loss_percentage = (lost / sent) * 100

    print("\nPackets Lost :", lost)
    print("Packet Loss : {:.2f}%".format(loss_percentage))

    if loss_percentage <= 2:
        print("Network Quality : Excellent")
    elif loss_percentage <= 5:
        print("Network Quality : Good")
    else:
        print("Network Quality : Poor")
```

```python
import pandas as pd

# Read the uploaded CSV file
data = pd.read_csv("bugs.csv")

# Display the dataset
data
```

```python
import pandas as pd

# Enter the number of bugs reported on each day
bug_counts = [
    2, 4, 3, 5, 1,
    3, 4, 2, 6, 3,
    5, 4, 2, 3, 7,
    4, 3, 5, 2, 4
]

# Create the day numbers automatically
days = range(1, len(bug_counts) + 1)

# Create a DataFrame
data = pd.DataFrame({
    "Day": days,
    "Bugs": bug_counts
})

# Display the dataset
data
```

```python
# Display some basic information about the bug data

print("Number of days:", len(data))
print("Average bugs per day:", data["Bugs"].mean())
print("Minimum bugs in a day:", data["Bugs"].min())
print("Maximum bugs in a day:", data["Bugs"].max())
```

```python
import matplotlib.pyplot as plt

# Plot the number of bugs reported each day
plt.figure(figsize=(10, 5))

plt.plot(
    data["Day"],
    data["Bugs"],
    marker="o"
)

plt.xlabel("Day")
plt.ylabel("Number of bugs")
plt.title("Daily Software Bug Reports")

plt.grid(True)
plt.show()
```

```python
from scipy.stats import poisson
```

```python
# Estimate the average number of bugs reported per day

lambda_value = data["Bugs"].mean()

print("Estimated bug rate (lambda):", lambda_value)
```

```python
# Probability of exactly 5 bugs in a day

probability_5_bugs = poisson.pmf(
    5,
    lambda_value
)

print("Probability of exactly 5 bugs:", probability_5_bugs)
```

```python
print(
    f"Probability of exactly 5 bugs: "
    f"{probability_5_bugs * 100:.2f}%"
)
```

```python
# Probability of having at most 3 bugs

probability_at_most_3 = poisson.cdf(
    3,
    lambda_value
)

print(
    f"Probability of at most 3 bugs: "
    f"{probability_at_most_3 * 100:.2f}%"
)
```

The development team considers a day with more than 6 bugs to be a high-bug day.
What is the probability that tomorrow will be a high-bug day?

```python
# Probability of getting more than 6 bugs
# We calculate this as:
# 1 - probability of getting 6 or fewer bugs

probability_high_bug_day = 1 - poisson.cdf(
    6,
    lambda_value
)

print(
    f"Probability of a high-bug day: "
    f"{probability_high_bug_day * 100:.2f}%"
)
```

```python
# Possible numbers of bugs we want to investigate
bug_numbers = range(0, 11)

# Calculate the probability of each possible bug count
probabilities = poisson.pmf(
    list(bug_numbers),
    lambda_value
)

plt.figure(figsize=(10, 5))

plt.bar(
    bug_numbers,
    probabilities
)

plt.xlabel("Number of bugs in a day")
plt.ylabel("Probability")
plt.title("Poisson Model of Daily Software Bugs")

plt.show()
```
